## 1. ¿Qué hace el mecanismo de atención y por qué son importantes query, key y value?

El mecanismo de atención es el componente que permite a un Transformer identificar qué partes del contexto son más relevantes al procesar o generar cada token. En lugar de considerar todas las palabras con la misma importancia, el modelo calcula una puntuación de relevancia entre el token actual y las demás posiciones de la secuencia. Para ello utiliza tres vectores: **Query (Q)** representa la información que el token actual busca; **Key (K)** describe el contenido disponible en cada posición; y **Value (V)** almacena la información que puede ser recuperada. La similitud entre Query y Key determina los pesos de atención, y esos pesos se aplican sobre los Value para construir una representación contextualizada. Gracias a este mecanismo, el modelo puede capturar relaciones entre palabras cercanas y lejanas, mejorando la comprensión del contexto.

### Respuesta

La salida anterior muestra el comportamiento de un **modelo base**, cuyo objetivo principal es predecir el siguiente token a partir del contexto previo. Este tipo de modelo continúa el texto de forma estadística, pero no necesariamente interpreta una instrucción ni responde con un formato útil para el usuario. En cambio, un **modelo alineado** ha sido ajustado mediante entrenamiento con instrucciones y retroalimentación humana para seguir órdenes, respetar formatos y priorizar respuestas útiles y seguras. Por ello, en tareas como clasificación, extracción de información o asistencia conversacional, un modelo alineado resulta mucho más práctico. La principal diferencia no es el conocimiento del modelo, sino su capacidad para comprender la intención del usuario y adaptar su respuesta a una tarea específica.

## 3. ¿Cuándo vale la pena pagar el costo de Chain-of-Thought?

En la Parte 3 del taller observé que **Chain-of-Thought** alcanzó una exactitud similar al **Zero-shot**, pero generó una cantidad mucho mayor de tokens de salida. Esto demuestra que el razonamiento explícito tiene un costo adicional en latencia y consumo de tokens que no siempre se traduce en una mejora del rendimiento. Considero que vale la pena utilizar Chain-of-Thought cuando la tarea requiere varios pasos de inferencia, verificación de restricciones o explicaciones detalladas, como problemas matemáticos, lógica o análisis complejos. En cambio, para tareas sencillas de clasificación o respuestas directas, Zero-shot ofrece una mejor relación entre costo y desempeño. La decisión debe depender de la complejidad del problema y no asumir que más razonamiento siempre produce mejores resultados.

## 4. ¿Qué gana y qué pierde un proveedor al retirar un parámetro de muestreo?

Durante la Parte 2 comprobé que GPT-5.6 rechazó parámetros como `temperature` y `top_p`, mientras que otros modelos sí los aceptaban. Al retirar estos parámetros, el proveedor obtiene una interfaz más simple y consistente, reduce configuraciones incompatibles y puede optimizar internamente el comportamiento del modelo sin depender de controles heredados. Sin embargo, el usuario pierde una herramienta importante para controlar la diversidad y la aleatoriedad de la generación. En la práctica, ese control suele trasladarse hacia mecanismos de mayor nivel, como modelos especializados o niveles de esfuerzo de razonamiento. Este cambio simplifica el uso de la API, pero también limita la capacidad del desarrollador para ajustar con precisión el comportamiento del modelo en aplicaciones específicas.

## 5. ¿Los reasoning tokens constituyen una explicación fiel del razonamiento del modelo?

No necesariamente. Los **reasoning tokens** representan el cómputo utilizado durante el proceso de inferencia, pero no deben interpretarse como una descripción exacta del razonamiento interno del modelo. En la Parte 4 del taller utilicé el contador de reasoning tokens para medir el nivel de esfuerzo, observando que incluso cuando el parámetro era aceptado, algunos casos registraban cero tokens de razonamiento. Esto demuestra que el contador es una métrica objetiva de uso, pero no una explicación completa de cómo se obtuvo la respuesta. Como argumenta Lanham et al. (2023), las cadenas de razonamiento visibles pueden ser explicaciones generadas por el propio modelo y no un registro fiel de sus procesos internos. Por ello, la evaluación debe apoyarse en métricas reproducibles y resultados verificables.
